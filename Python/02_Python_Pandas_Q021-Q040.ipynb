##32. Use IQR to flag price outliers.

```python
IQR_cal = erp_order_items[['product_id','unit_price']].groupby('product_id').describe()
IQR_cal.head(10)
IQR_cal['IQR'] = IQR_cal[('unit_price','75%')]-IQR_cal[('unit_price','25%')]
IQR_cal['upper']=IQR_cal[('unit_price','75%')]+(1.5*IQR_cal['IQR'])
IQR_cal['lower']=IQR_cal[('unit_price','25%')]-(1.5*IQR_cal['IQR'])
IQR_cal.head(10)
```

```python
IQR_cal.reset_index(inplace=True)
```

```python
bounds.head(10)
```

```python
bounds.columns = ['product_id', 'upper', 'lower']
```

```python
bounds.head()
```

```python
erp_order_items = erp_order_items.merge(bounds,on='product_id',how='left')
```

##38. Create a reusable data-quality report.

```python
def data_quality_report(df):
    report = {
        'total_rows': len(df),
        'duplicate_rows': df.duplicated().sum(),
        'missing_values': df.isna().sum().sum(),
        'invalid_invoice_totals': (
            (df['subtotal'] + df['tax_amount']).round(2)
            != df['invoice_total'].round(2)
        ).sum(),
        'reversed_invoices': (
            df['invoice_status'] == 'Reversed'
        ).sum(),
        'duplicate_invoice_keys': df.duplicated(
            subset=['invoice_id', 'order_id', 'invoice_date']
        ).sum()
    }

    return pd.DataFrame(
        report.items(),
        columns=['check', 'count']
    )

    
```

```python
data_quality_report(finance_invoices)
```

##40. Write functions for common validation checks.

```python
finance_invoices = pd.read_csv(
    r"F:\Project Data Master\Asteron_Data_Lab_v1_Learner_Pack\raw\finance_invoices.csv"
)
```

```python
finance_invoices.head()
```

```python
def validation_check(df):
    valid_total = (
        df['invoice_total'].round(2) ==
        (df['subtotal'] + df['tax_amount']).round(2)
    )

    result = {
        'Duplicate Values': df.duplicated().sum(),
        'Missing values': df.isna().sum().sum(),
        'Unique key': df['invoice_id'].nunique() == len(df),
        'Valid invoice total': valid_total.sum(),
        'Invalid invoice total': (~valid_total).sum(),
        'Reversed invoices': (df['invoice_status'] == 'Reversed').sum(),
        'Posted invoices': (df['invoice_status'] == 'Posted').sum()
    }

    return pd.DataFrame(result.items(), columns=['Check', 'Count'])
```

```python
validation_check(finance_invoices)
```
